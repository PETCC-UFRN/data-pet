```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
try:
    path = '../../2.0-dados-abertos-UFRN/1-limpeza-de-dados/databases/turmas_bti.csv'
    df = pd.read_csv(path, sep=';')
except Exception as e:
    print(f"Erro ao carregar dados: {e}")
```

```python
plt.figure(figsize=(18, 18))

plt.subplot(3, 2, 1)
sns.histplot(df['ch_dedicada_periodo'], kde=True, bins=20, color='skyblue')
plt.title('1. Distribuição de Carga Horária Dedicada')
plt.xlabel('CH Dedicada por Período')

plt.subplot(3, 2, 2)
sns.histplot(df['capacidade_aluno'], kde=True, bins=20, color='salmon')
plt.title('2. Distribuição de Capacidade de Alunos')
plt.xlabel('Capacidade de Alunos')
```

```python
plt.figure(figsize=(18, 18))
plt.subplot(3, 2, 3)
a = sns.scatterplot(x='capacidade_aluno', y='total_solicitacoes', data=df, alpha=0.6)
plt.title('3. Capacidade vs. Total de Solicitações')
plt.xlabel('Capacidade de Alunos')
plt.ylabel('Total de Solicitações')
```

```python
#a.ax.axline(xy1=(10,2), slope=.5, color="b")
b = sns.relplot(data=df, x='capacidade_aluno', y='total_solicitacoes', s=23, color='palevioletred', alpha=0.6)
b.ax.axline(xy1=(1, 1), xy2=(2,2), alpha=0.6, color="purple")
```

```python
plt.figure(figsize=(18, 18))
plt.subplot(3, 2, 4)
sns.countplot(y='situacao_turma', data=df, palette='Spectral', order=df['situacao_turma'].value_counts().index)
plt.title('4. Contagem por Situação da Turma')
plt.xlabel('Contagem')
plt.ylabel('Situação da Turma')
```

```python
plt.figure(figsize=(18, 18))
plt.subplot(3, 2, 5)
sns.countplot(x=df['ano'].astype(str), hue='periodo', data=df, palette='viridis')
plt.title('5. Número de Turmas por Ano e Período')
plt.xlabel('Ano')
plt.ylabel('Contagem de Turmas')
plt.xticks(rotation=45, ha='right')
```

```python
plt.figure(figsize=(18, 18))
top_disciplinas = df['nome'].value_counts().head(20)
plt.subplot(3, 2, 6)
sns.barplot(x=top_disciplinas.values, y=top_disciplinas.index, palette='magma')
plt.title('6. Rank das 20 Componentes com mais Turmas')
plt.xlabel('Número de Turmas')
plt.ylabel('Nome do Componente')
```
