```python
import os
import pandas as pd
```

```python
componentes_df=pd.read_csv("data/componentes.csv")
turmas_df=pd.read_csv("data/turmas.csv")
```

```python
turmas_df['id_componente'] = turmas_df['id_componente_curricular']
turmas_df = turmas_df.drop(columns=["id_componente_curricular"])
```

```python
turmas_completas = pd.merge(
    turmas_df, 
    componentes_df, 
    on='id_componente', 
    how='left'
)
```

```python
arquivo_saida = 'dados_completos_ufrn.csv'
chunk_size = 50000
chunks = pd.read_csv("data/matriculas.csv", chunksize=chunk_size)
first_chunk = True
```

```python
for chunk_matriculas in chunks:
    # Faz o merge apenas deste pedaço
    merge_parcial = pd.merge(
        chunk_matriculas, 
        turmas_completas, 
        on='id_turma', 
        how='left'
    )
    
    # Salva no disco imediatamente (mode='a' significa append/adicionar ao final)
    # Se for o primeiro pedaço, escreve o cabeçalho (header). Nos próximos, não.
    merge_parcial.to_csv(
        arquivo_saida, 
        mode='w' if first_chunk else 'a', 
        header=first_chunk, 
        index=False,
        sep=';'
    )
    
    first_chunk = False
    
    # Dica extra: Força a limpeza da memória a cada ciclo
    del merge_parcial
```

```python
print("Processo finalizado! O arquivo gigante foi criado no disco.")
```

```python
# Verificando Componentes
# Pelo dicionário, a chave é 'id_componente' [cite: 41]
duplicadas_comp = componentes_df['id_componente'].duplicated().sum()
print(f"Componentes duplicados: {duplicadas_comp}")

# Verificando Turmas
# Pelo dicionário, a chave é 'id_turma' [cite: 10]
duplicadas_turma = turmas_df['id_turma'].duplicated().sum()
print(f"Turmas duplicadas: {duplicadas_turma}")
```
