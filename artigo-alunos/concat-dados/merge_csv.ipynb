```python
import os
import pandas as pd
```

```python
matriculas_df=pd.read_csv("data/matriculas.csv")
componentes_df=pd.read_csv("data/componentes.csv")
turmas_df=pd.read_csv("data/turmas.csv")
```

```python
turmas_df['id_componente'] = turmas_df['id_componente_curricular']
turmas_df = turmas_df.drop(columns=["id_componente_curricular"])
```

```python
turmas_completas_df = pd.merge(
    left=turmas_df,
    right=componentes_df,
    on='id_componente',
    how='left'
)
turmas_completas_df.to_csv("turmas_completas.csv",index=False)
```

```python
df_final = pd.merge(
    left=matriculas_df,
    right=turmas_completas_df,
    on='id_turma',
    how='left'
)
```

```python
df_final.to_csv("dataset_matriculas_componentes_detalhado.csv", index=False)
```
