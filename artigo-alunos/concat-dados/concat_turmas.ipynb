```python
import os
path_to_csv = os.path.join("data", "turmas-dataset")
```

```python
import glob
import pandas as pd
```

```python
if not os.path.exists(path_to_csv):
    raise ValueError(f"O diretório {path_to_csv} não existe.")
csv_files = glob.glob(os.path.join(path_to_csv, "*.csv"))
```

```python
df_list=[]
if not csv_files:
    print("Nenhum arquivo CSV encontrado no diretório.")
for file in csv_files:
    df_list.append(pd.read_csv(file, sep=";"))
concatenated_df = pd.concat(df_list, ignore_index=True)
print(f"Removendo {int(concatenated_df.duplicated().sum())} linhas duplicadas do dataset")
concatenated_df = concatenated_df.drop_duplicates()
output_path = os.path.join("data", "turmas.csv")
concatenated_df.to_csv(output_path, index=False)
print(f"Arquivos concatenados e salvos em: {output_path}")
```
