```python
from pathlib import Path

csv_dir = Path('../discentes_csv')
files = sorted(csv_dir.glob('*.csv'))

if not files:
    raise FileNotFoundError(f'No CSV files found in {csv_dir!s}')

dfs = []
for f in files:
    try:
        # tenta detectar separador automaticamente
        dfs.append(pd.read_csv(f, sep=None, engine='python'))
    except Exception:
        # fallback para ; caso muitos arquivos usem ponto-e-vírgula
        dfs.append(pd.read_csv(f, sep=';'))

df_discentes = pd.concat(dfs, ignore_index=True)

print(f'Concatenated {len(files)} files -> df_discentes.shape = {df_discentes.shape}')
```

```python
sample_names = ["FERNANDO SAMPAIO DO CARMO",
"ALYSSON ZENILDO COSTA ALVES",              
"HERBENNY HELLEN MEDEIROS DE BRITO CARVALHO",
"MONIQUE MAIA DE LIMA",
"ARTHUR FELIPE RODRIGUES DE LIMA",
"MARIANA PEREIRA PATUCCI",
"SAMARA KELLY SOUZA DE OLIVEIRA",
"ISABELA MARIZ DE OLIVEIRA",
"ELIEZER ROCHA DA SILVA",
"GILMAR ALEXANDRE GUEDES JUNIOR"]
```

```python
# seleciona e imprime todas as linhas em df_discentes cujos nome_discente estão em sample_names
mask = df_discentes['nome_discente'].isin(sample_names)
selected_rows = df_discentes[mask]

print(f'Selected rows: {selected_rows.shape[0]}')
print(selected_rows.to_string(index=False))
```
