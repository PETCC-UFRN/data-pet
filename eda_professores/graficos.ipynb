## Verificando se todos os dados de df['lotacao'] estão presentes em listas anteriores (de centros ou departamentos)

```python
todos_deptos = df['lotacao'].unique().tolist()

centros = [cb, ccs, ccet, cchla, ccsa, ceres, ce, ct, eaj, ect, emufrn, es, emcm, facisa, felcs, imd, nei, iq, ice, sti, agora, iif, ipp, ih, outros]

deptos_presentes = []
for lista in centros:
    deptos_presentes.extend(lista)
    
deptos_presentes = list(set(deptos_presentes))

deptos_ausentes = [depto for depto in todos_deptos if depto not in deptos_presentes]

total_deptos = len(todos_deptos)
total_presente = len(deptos_presentes)
total_ausente = len(deptos_ausentes)

df_ausentes = (
    df[df['lotacao'].isin(deptos_ausentes)]  
    .groupby('lotacao')['nome'].nunique()    
    .reset_index(name='profs_vinculados')    
    .sort_values('profs_vinculados', ascending=False)  
)


print("\n=== DEPARTAMENTOS AUSENTES E PROFESSORES ===")
print(df_ausentes)

print("\n=== RESUMO ===")
print(f"Total de departamentos únicos: {total_deptos}")
print(f"Departamentos presentes nas listas: {total_presente} ({', '.join(deptos_presentes)})")
print(f"Departamentos ausentes nas listas: {total_ausente} ({', '.join(deptos_ausentes)})")

print("\n=== DETALHES ===")
print(f"Lista completa de departamentos: {todos_deptos}")
print(f"Departamentos considerados presentes: {deptos_presentes}")
print(f"Departamentos faltantes: {deptos_ausentes}")
```

### Métricas dos profs de exatas (2013-2024)

```python
import plotly.express as px

df_exatas['media_final_geral'] = df_exatas[
    ['postura_profissional_media', 'atuacao_profissional_media', 'autoavaliacao_aluno_media']
].mean(axis=1)


df_exatas['ano'] = df_exatas['ano'].astype(int)
df_exatas = df_exatas[(df_exatas['ano'] >= 2013) & (df_exatas['ano'] <= 2024)]

# agrupando por ano e depto
df_metrics = df_exatas.groupby(['ano', 'lotacao'])[
    ['postura_profissional_media', 'atuacao_profissional_media', 
     'autoavaliacao_aluno_media', 'media_final_geral']
].mean().reset_index()

# gráfico
fig = px.line(
    df_metrics,
    x='ano',
    y=['postura_profissional_media', 'atuacao_profissional_media', 
       'autoavaliacao_aluno_media', 'media_final_geral'],
    color='lotacao',
    title='Métricas dos Departamentos de Exatas (2013-2024)',
    labels={'value': 'Média', 'variable': 'Métrica', 'lotacao': 'Departamento'},
    line_dash='variable',
    facet_col='variable',
    facet_col_wrap=2,
    height=600,
    width=1000
)

#ajustes
fig.update_layout(
    hovermode='x unified',
    legend_title='Departamento/Métrica',
    xaxis_title='Ano',
    yaxis_title='Média'
)

fig.show()
```
