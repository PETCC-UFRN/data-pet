```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
plt.style.use('ggplot')
```

```python
df = pd.read_csv("../limpezaSegundafase/CSVs/Finais/avaliacoes_Docentes_Turmas.csv", sep = ";")
```

#### Análise com base no gênero

Para analisar o dataset com base nessa métrica, vê-se necessária a criação de dataframes específicos para os sexos F e M

```python
df_professores_m = df[df['sexo'] == 'M']
df_professores_f = df[df['sexo'] == 'F']
```

Agora, com base nos dois dataframes gerados, podemos comparar o desempenho dos profissionais, das suas turmas e das autoavaliações dos seus alunos

##### Sexo Masculino

```python
notas = ['postura_profissional_media', 'atuacao_profissional_media', 'autoavaliacao_aluno_media', 'ano_admissao', 'media_final_geral']
fig, axes = plt.subplots(3, 2, figsize=(12, 10))
axes = axes.flatten()

for i, notas in enumerate(notas):
    sns.histplot(df_professores_m[notas], bins=30, kde=True, color='skyblue', ax=axes[i])
    axes[i].set_title(f'Distribuição dos valores: {notas}')
    axes[i].set_xlabel('Valores')
    axes[i].set_ylabel('Frequência')

# Remove o subplot extra (o 6º)
fig.delaxes(axes[-1])

plt.tight_layout()
plt.show()
```

##### Sexo Feminino

```python
areas = ['postura_profissional_media', 'atuacao_profissional_media', 'autoavaliacao_aluno_media', 'ano_admissao', 'media_final_geral']
fig, axes = plt.subplots(3, 2, figsize=(12, 10))
axes = axes.flatten()

for i, area in enumerate(areas):
    sns.histplot(df_professores_f[area], bins=30, kde=True, color='skyblue', ax=axes[i])
    axes[i].set_title(f'Distribuição dos valores: {area}')
    axes[i].set_xlabel('Valores')
    axes[i].set_ylabel('Frequência')

# Remove o subplot extra (o 6º)
fig.delaxes(axes[-1])
#fig.delaxes(axes[-2])

plt.tight_layout()
plt.show()
```
