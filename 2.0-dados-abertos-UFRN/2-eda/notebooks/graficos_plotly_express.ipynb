```python
df_exatas['media_final_geral'] = df_exatas[
    ['postura_profissional_media', 'atuacao_profissional_media', 'autoavaliacao_aluno_media']
].mean(axis=1)


df_exatas['ano'] = df_exatas['ano'].astype(int)
df_exatas = df_exatas[(df_exatas['ano'] >= 2013) & (df_exatas['ano'] <= 2024)]

# agrupando por ano e depto
df_metrics = df_exatas.groupby(['ano', 'lotacao'])[
    ['postura_profissional_media', 'atuacao_profissional_media', 
     'autoavaliacao_aluno_media', 'media_final_geral']
].mean().reset_index()

# gráfico
fig = px.line(
    df_metrics,
    x='ano',
    y=['postura_profissional_media', 'atuacao_profissional_media', 
       'autoavaliacao_aluno_media', 'media_final_geral'],
    color='lotacao',
    title='Métricas dos Departamentos de Exatas (2013-2024)',
    labels={'value': 'Média', 'variable': 'Métrica', 'lotacao': 'Departamento'},
    line_dash='variable',
    facet_col='variable',
    facet_col_wrap=2,
    height=600,
    width=1000
)

#ajustes
fig.update_layout(
    hovermode='x unified',
    legend_title='Departamento/Métrica',
    xaxis_title='Ano',
    yaxis_title='Média'
)

fig.write_html('graficos/metricas_deptos_exatas.html')
```

```python
# cálculo das médias anuais
df_yearly_means = df.groupby('ano').agg(
    postura_profissional_media_anual=('postura_profissional_media', 'mean'),
    atuacao_profissional_media_anual=('atuacao_profissional_media', 'mean'),
    autoavaliacao_aluno_media_anual=('autoavaliacao_aluno_media', 'mean')
).reset_index()

# transformação para o plotly
df_yearly_means_long = pd.melt(df_yearly_means,
                               id_vars=['ano'],
                               value_vars=['postura_profissional_media_anual',
                                           'atuacao_profissional_media_anual',
                                           'autoavaliacao_aluno_media_anual'],
                               var_name='metrica',
                               value_name='media_avaliacao')

fig_evolucao_temporal = px.line(df_yearly_means_long,
                                x='ano',
                                y='media_avaliacao',
                                color='metrica',
                                title='Evolução Temporal das Médias de Avaliação (2013-2024)',
                                labels={'ano': 'Ano', 'media_avaliacao': 'Média da Avaliação', 'metrica': 'Métrica de Avaliação'},
                                markers=True)
fig_evolucao_temporal.update_layout(xaxis_type='category')
#fig_evolucao_temporal.show()

fig_evolucao_temporal.write_html("graficos/evolucao_temporal_avaliacoes.html")
```
