```python


# 1. Configurar a figura e os eixos (subplots)
# Vamos criar uma grade de 3 linhas por 2 colunas para caber os 5 gráficos.
# figsize=(14, 12) define o tamanho total da imagem.
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(14, 12))

# 2. Achatar o array de eixos
# 'axes' é um array 2D (3x2). 'axes.flatten()' o transforma em um array 1D (comprimento 6),
# o que facilita iterar por ele (ax[0], ax[1], ax[2], ...)
ax = axes.flatten()


# --- Gráfico 1: ano_ingresso (Gráfico de Barras) ---
# Usamos .value_counts() para contar as ocorrências de cada ano
# e .sort_index() para garantir que os anos fiquem em ordem (2013, 2014, ...)
ano_counts = df_discentes['ano_ingresso'].value_counts().sort_index()

# Criamos o gráfico de barras no primeiro eixo (ax[0])
ax[0].bar(ano_counts.index.astype(str), ano_counts.values, color='skyblue', edgecolor='black')
ax[0].set_title('Distribuição por Ano de Ingresso')
ax[0].set_xlabel('Ano')
ax[0].set_ylabel('Contagem (Nº de Discentes)')
# Rotaciona os labels do eixo X para melhor legibilidade
ax[0].tick_params(axis='x', rotation=45)


# --- Gráfico 2: periodo_ingresso (Gráfico de Barras) ---
# Mesmo processo do anterior, mas para 'periodo_ingresso'
periodo_counts = df_discentes['periodo_ingresso'].value_counts().sort_index()

# Criamos o gráfico de barras no segundo eixo (ax[1])
# Usamos .astype(str) para garantir que matplotlib trate os períodos (0.0, 1.0) como categorias
ax[1].bar(periodo_counts.index.astype(str), periodo_counts.values, color='salmon', edgecolor='black')
ax[1].set_title('Distribuição por Período de Ingresso')
ax[1].set_xlabel('Período')
ax[1].set_ylabel('Contagem (Nº de Discentes)')


# --- Gráfico 3: id_curso (Histograma) ---
# Para 'id_curso', um histograma é mais apropriado.
# .dropna() é importante para remover valores ausentes (NaN) que o describe() indicou
# bins=50 significa que queremos agrupar os dados em 50 "caixas"
ax[2].hist(df_discentes['id_curso'].dropna(), bins=50, color='lightgreen', edgecolor='black')
ax[2].set_title('Distribuição de id_curso')
ax[2].set_xlabel('ID do Curso')
ax[2].set_ylabel('Frequência')
# Usar notação científica no eixo X para IDs grandes
ax[2].ticklabel_format(style='sci', axis='x', scilimits=(0,0))


# --- Gráfico 4: id_unidade (Histograma) ---
# Mesmo processo para 'id_unidade'
ax[3].hist(df_discentes['id_unidade'].dropna(), bins=50, color='gold', edgecolor='black')
ax[3].set_title('Distribuição de id_unidade')
ax[3].set_xlabel('ID da Unidade')
ax[3].set_ylabel('Frequência')
# AVISO: Pelos seus dados (std > mean), este gráfico pode ser muito distorcido (skewed).
# Se a maioria dos valores estiver em uma única barra,
# tente adicionar: ax[3].set_yscale('log') para usar escala logarítmica.


# --- Gráfico 5: id_unidade_gestora (Histograma) ---
# Mesmo processo para 'id_unidade_gestora'
ax[4].hist(df_discentes['id_unidade_gestora'].dropna(), bins=50, color='orchid', edgecolor='black')
ax[4].set_title('Distribuição de id_unidade_gestora')
ax[4].set_xlabel('ID da Unidade Gestora')
ax[4].set_ylabel('Frequência')


# --- Limpar o eixo extra ---
# Nossa grade é 3x2 (6 gráficos), mas só usamos 5.
# Vamos desligar o último eixo (ax[5])
ax[5].axis('off')

# 3. Ajustar o layout e salvar
# plt.tight_layout() ajusta automaticamente os gráficos para que não haja sobreposição de títulos/labels
plt.tight_layout()

# 4. Salvar a figura
# Salva a imagem completa em um arquivo PNG
output_filename = 'distribuicao_variaveis_discentes.png'
plt.savefig(output_filename)

print(f"Gráficos salvos em {output_filename}")

# Fechamos a figura para liberar memória
plt.close()
```

```python
colunas_categoricas = [x for x in df_discentes.columns if df_discentes[x].dtype == 'object']
print(colunas_categoricas)
```

```python

print("\n--- Análise de Cardinalidade (Valores Únicos) ---")

# Dicionário para guardar a contagem de valores únicos
unique_counts = {}

# Itere pelas colunas e pegue a contagem de únicos
for col in colunas_categoricas:
    n_unique = df_discentes[col].nunique()
    unique_counts[col] = n_unique
    print(f"- Coluna '{col}': {n_unique} valores únicos")

print("--------------------------------------------------")
print("Análise de cardinalidade concluída.")
print("Com base nisso, podemos decidir a melhor forma de plotar cada variável.")
```

```python

# --- Listas de Colunas (sem alteração) ---
colunas_baixa_cardinalidade = [
    'sexo', 
    'tipo_discente', 
    'status', 
    'sigla_nivel_ensino', 
    'nivel_ensino', 
    'modalidade_educacao',
    'nome_unidade_gestora'
]

colunas_alta_cardinalidade = [
    'forma_ingresso', 
    'nome_curso', 
    'nome_unidade'
]

N_TOP = 15
print(f"Gerando {len(colunas_baixa_cardinalidade)} gráficos para baixa cardinalidade...")

# --- Loop 1: Gráficos de Baixa Cardinalidade (um por um) ---
for col in colunas_baixa_cardinalidade:
    # 1. Criar UMA NOVA figura para ESTE gráfico
    # Um tamanho 10x7 é bom para um gráfico vertical
    fig, ax = plt.subplots(figsize=(10, 7))
    
    # 2. Calcular os dados (sem alteração)
    counts = df_discentes[col].dropna().value_counts()
    
    # 3. Plotar no eixo 'ax' (não mais 'ax[i]')
    ax.bar(counts.index.astype(str), counts.values, color='c', edgecolor='black')
    ax.set_title(f'Distribuição de {col} (Todos)')
    ax.set_ylabel('Contagem')
    ax.tick_params(axis='x', rotation=45)
    
    # 4. Ajustar layout para esta figura
    plt.tight_layout()
    
    # 5. Criar um nome de arquivo dinâmico e salvar
    output_filename = f'distribuicao_{col}.png'
    plt.savefig(output_filename)
    print(f"- Imagem salva: {output_filename}")
    
    # 6. Fechar a figura atual para liberar memória
    plt.close(fig)

print(f"\nGerando {len(colunas_alta_cardinalidade)} gráficos Top {N_TOP} para alta cardinalidade...")

# --- Loop 2: Gráficos de Alta Cardinalidade (um por um) ---
for col in colunas_alta_cardinalidade:
    # 1. Criar UMA NOVA figura
    # 12x8 é bom para barras horizontais (mais espaço para labels)
    fig, ax = plt.subplots(figsize=(12, 8))
    
    # 2. Calcular os dados (sem alteração)
    counts = df_discentes[col].dropna().value_counts().nlargest(N_TOP)
    total_unicos = df_discentes[col].nunique()

    # 3. Plotar no eixo 'ax' (não mais 'ax[i]')
    ax.barh(counts.index.astype(str), counts.values, color='m', edgecolor='black')
    ax.set_title(f'Top {N_TOP} para {col} (de {total_unicos} únicos)')
    ax.set_xlabel('Contagem')
    ax.invert_yaxis() # Mais frequente no topo
    
    # 4. Ajustar layout
    plt.tight_layout()
    
    # 5. Criar nome de arquivo e salvar
    output_filename = f'distribuicao_top15_{col}.png'
    plt.savefig(output_filename)
    print(f"- Imagem salva: {output_filename}")
    
    # 6. Fechar a figura
    plt.close(fig)

print("\nProcesso concluído. Todas as 10 imagens foram geradas.")
```
