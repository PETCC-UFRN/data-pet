# Tratamento dos datasets [Situações de Discentes](https://dados.ufrn.br/dataset/situacoes-de-discentes)

O presente notebook aplica o processo de tratamento aos datasets de situações de discentes dos anos de 2013 à 2024.

Ao realizar os downloads dos .csvs, os datasets eram salvos com a extensão de arquivo .1 e .2 referentes ao período letivo dos respectivos anos (por exemplo, o dataset de 2013.1 era salvo como 'situacoes-discentes-2013.1' em vez de 'situacoes-discentes-2013-1.csv').

Desse modo, alterei manualmente o nome dos arquivos para 'situacoes-discentes-20XX-X.csv'.

Abaixo, estão dispostas as células para o tratamento destes .csv's e algumas análises superficiais para justificar a limpeza do dataframe.

OBS.: Carregarei no repositório APENAS A VERSÃO FINAL de 'situacoes-discentes', com dados de 2013 a 2024 referentes ao nível de ensino "GRADUAÇÃO".

```python
import pandas as pd
import glob
import os
```

```python
arquivos = os.path.join("../situacoes_discentes/situacoes-discentes-20*.csv")

lista_arquivos = sorted(glob.glob(arquivos))

for f in lista_arquivos:
    print(f"  - {os.path.basename(f)}")
```

```python
lista_df = []

for arquivo in lista_arquivos:
    df_temp = pd.read_csv(arquivo, sep=None, engine='python')
    lista_df.append(df_temp)
    
df_concat = pd.concat(lista_df, ignore_index=True)

#df_concat = df_concat.to_csv("../databases/situacoes-discentes.csv", sep=';', index=False)

df_concat
```

```python
df_concat.info()
```

* De 2013.1 até 2024.2, temos 1.073.388 linhas no dataset
* Não há observação de valores faltantes
* Não é relevante para nossa análise manter a coluna 'data_alteracao_situacao'
* Atualmente, não é relevante para nossa análise manter discentes com nível de ensino diferente de graduação.

```python
#remoção da data de alteração da situação do discente
df_concat = df_concat.drop(columns=['data_alteracao_situacao'])
#recorte da análise para discentes de graduação
df_concat = df_concat[df_concat['nivel_ensino'] == "GRADUAÇÃO"]

df_concat.info()
```

* De 1.073.388 valores, agora temos 951.714

```python
df_concat = df_concat.to_csv("../databases/situacoes-discentes.csv", sep=';', index=False)
```
