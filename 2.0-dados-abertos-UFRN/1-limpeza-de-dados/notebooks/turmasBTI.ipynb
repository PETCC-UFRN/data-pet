# Criando o DataFrame de Turmas do BTI entre 2013 e 2024

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import glob
import os
```

```python
#carregando o dataframe principal
turmas_bti = pd.read_csv('../databases/turmas.csv')
#carregando o dataframe secundário, vamos usar algumas colunas dele 
comp_curr_bti = pd.read_csv('../databases/comp_curr_BTI.csv')
```

```python
#visualização do df
turmas_bti
```

```python
#dropando colunas irrelevantes p/ nossa análise
turmas_bti = turmas_bti.drop(['matricula_docente_externo', 'observacao', 'campus_turma', 'local', 'data_inicio', 'data_fim', 'distancia', 'data_consolidacao', 'agrupadora', 'id_turma_agrupadora', 'convenio', 'modalidade_participantes'], axis=1)

turmas_bti
```

```python
#apenas turmas de graduação
turmas_bti = turmas_bti[turmas_bti['nivel_ensino'] == 'GRADUAÇÃO']
#apenas turmas regulares
turmas_bti = turmas_bti[turmas_bti['tipo'] == 'REGULAR']

turmas_bti
```

```python
#filtrando id_componente_curricular para manter no df apenas componentes curriculares ofertados ao bti
turmas_bti = turmas_bti[turmas_bti['id_componente_curricular'].isin(comp_curr_bti['id_componente'])]

turmas_bti
```

```python
#unindo colunas 'id_componente', 'codigo' e 'nome' de comp_curr_bti no df turmas_bti, join na coluna de 'id_componente'
turmas_bti = turmas_bti.merge(comp_curr_bti[['id_componente', 'codigo', 'nome']], left_on='id_componente_curricular', right_on='id_componente', how='left')

turmas_bti
```

```python
#dropando outras colunas irrelevantes
turmas_bti.drop(['id_componente', 'tipo', 'nivel_ensino', 'qtd_aulas_lancadas'], axis=1)
#reorganizando a ordem das colunas
turmas_bti = turmas_bti[['id_turma', 'codigo_turma', 'id_componente', 'codigo', 'nome', 'siape', 'ch_dedicada_periodo', 'ano', 'periodo', 'descricao_horario', 'situacao_turma', 'capacidade_aluno', 'total_solicitacoes']]

turmas_bti
```

```python
#exportando para csv com sep=';', prefiro assim !
turmas_bti.to_csv('../databases/turmas_bti.csv', sep=';')
```

Temos um dataset com turmas regulares ofertadas ao BTI (com código DIM ou IMD) entre 2013 e 2024.
